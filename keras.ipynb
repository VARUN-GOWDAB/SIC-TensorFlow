```python
import pandas as pd
import numpy as np
import os
import warnings
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam, RMSprop, SGD
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
df = pd.read_csv('TV_Sales.csv', header='infer', encoding = 'latin1')
X = df.drop(columns=['price'])
y = df['price']
```

```python
df.head(5)
```

```python
scaler = MinMaxScaler()
X = pd.get_dummies(
    X,
    columns=['shelve_loc', 'city', 'local'],
    dtype=float
)

X = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=123)
n_vars = X_train.shape[1]
```

```python
my_model1 = Sequential()
my_model1.add(Dense(input_dim = n_vars, units = 1, activation="linear"))

my_model1.summary()
```

```python
n_epochs = 2000
batch_size = 10
learn_rate = 0.002

my_optimizer = Adam(learning_rate=learn_rate)
my_model1.compile(loss = "mae", optimizer = my_optimizer, metrics=["mse"])
```
