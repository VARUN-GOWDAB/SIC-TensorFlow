```python
#Connecting two tensors - Approach 2
X = tf.constant([
    [80,75],[65,70],[90,88],[55,60]
])

y = tf.constant([1,0,1,0])

dataset = tf.data.Dataset.from_tensor_slices((X,y))

for item in dataset:
    print(item)
```

```python
ds_train = dataset.map(lambda x,y : (x* 2 - 1, y))

for X,y in ds_train:
    print(X.numpy(),y.numpy())
```

```python
def preprocess(x,y):
    x = x / 255.0
    return x,y

datas = dataset.map(preprocess)
```

```python
#shuffle, batch, repeat
#shuffle -> change the order/randomize
#batch -> group samples
#repeat -> go through dataset again

dataset = dataset.shuffle(1000)

for X,y in dataset:
    print(X.numpy(),y.numpy())
```

```python
import pathlib

imgdir_path = pathlib.Path("cat_dog_images")

file_list = sorted(
    [
        str(path)
        for path in imgdir_path.glob("*.jpg")
    ]
)

print(file_list)
```

```python
import matplotlib.pyplot as plt
import os

fig = plt.figure(figsize=(10, 5))

for i, file in enumerate(file_list):
    img_raw = tf.io.read_file(file) #raw binary data
    img = tf.image.decode_image(img_raw) #image tensor

    print("Image Size: ", img.shape) #H, W, RGB

    ax = fig.add_subplot(2, 3, i+1)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.imshow(img)
    ax.set_title(os.path.basename(file), size=15)

plt.tight_layout()
plt.show()
```

```python
import os

labels = [
    1 if "dog" in os.path.basename(file)
    else 0
    for file in file_list
]

print(labels)
```

```python
ds = tf.data.Dataset.from_tensor_slices((file_list, labels))

for X,y in ds:
    print(X.numpy(),y.numpy())
```

```python
def preprocess(path, label):
    image = tf.io.read_file(path)

    image = tf.image.decode_jpeg(image, channels=3)

    image = tf.image.resize(image, [80, 120])

    image = tf.cast(image, tf.float32) / 255.0

    return image, label
```

```python
ds = ds.map(preprocess)

for X,y in ds:
    print(X.numpy(),y.numpy())
```

```python
fig = plt.figure(figsize=(10, 5))

for i, example in enumerate(ds):
    print(example[0].shape, example[1].numpy())

    ax = fig.add_subplot(2, 3, i+1)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.imshow(example[0].numpy())

    ax.set_title("{}".format(example[1].numpy()), size=15)

plt.tight_layout()
plt.show()
```
