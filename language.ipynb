```python
from collections import Counter
import math
```

```python
corpus = [
    ["<s>", "I", "love", "machine", "learning", "</s>"],
    ["<s>", "I", "love", "deep", "learning", "</s>"],
    ["<s>", "machine", "learning", "is", "useful", "</s>"],
    ["<s>", "deep", "learning", "is", "useful", "</s>"],
    ["<s>", "I", "love", "machine", "learning", "</s>"]
]
```

```python
unigram_counts = Counter()

for sentence in corpus:
    unigram_counts.update(sentence)

total_tokens = sum(unigram_counts.values())

for word, count in unigram_counts.items():
    print(f"{word:10} : {count}")
```

```python
unigram_probs = {}

for word, count in unigram_counts.items():
    unigram_probs[word] = count / total_tokens

for word, prob in unigram_probs.items():
    print(f"{word:10} : {prob:.4f}")
```

```python
def unigram_prob_sent(sentence):
    probability = 1.0

    for word in sentence:
        if word not in unigram_probs:
            return 0.0

        probability *= unigram_probs[word]
    return probability
```

```python
sentence_A = ["<s>","I","love","machine","learning","</s>"]
sentence_B = ["<s>","I","love","deep","learning","</s>"]

unigram_A = unigram_prob_sent(sentence_A)
unigram_B = unigram_prob_sent(sentence_B)
```

```python
print(f"Unigram Approximation of A: {unigram_A:4f}")
print(f"Unigram Approximation of B: {unigram_B:4f}")
```

```python
from nltk.lm import MLE
from nltk.lm.preprocessing import padded_everygram_pipeline
```

```python
corpus = [
    ["I", "love", "machine", "learning"],
    ["I", "love", "deep", "learning"],
    ["machine", "learning", "is", "useful"],
    ["deep", "learning", "is", "useful"],
    ["I", "love", "machine", "learning"]
]

n = 2
```

```python
train_data, vocabulary = padded_everygram_pipeline(n, corpus)
```

```python
model = MLE(n) #empty

model.fit(train_data, vocabulary)
```

```python
# P(machine | love)

model.score("machine",["love"])
```
