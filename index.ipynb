```python
text = """
Hi my name is Huma and my number is 9876543210
My fathers number is 9012345678 and my mothers number
is 6543210987. You can also reach me at
7894561230
"""
pattern=r"\d{10}"
#check if entire text is a number? 
result=re.fullmatch(pattern,text)
print(result)
#find any phone number inside the sentence
result=re.search(pattern,text)
print(result)
#find all phone numbers inside the sentence
result=re.findall(pattern,text)
print(result)
#mask any phone number inside the sentence
result=re.sub(pattern,"*****",text)
print(result)
```

```python
pattern = r"([0-9]+)[^0-9]+([0-9]+)"
result = re.search(pattern,"Anna is 15 years old and John is 12 years old")
print(result.group(0))
print(result.group(1))
print(result.group(2))
```

```python
pattern = re.compile("([0-9]+)([^0-9]+[0-9]+)")
result = pattern.search(
    "Anna is 15 years old and John is 12 years old"
)
print(result.group(0))
print(result.group(1))
print(result.group(2))
```

```python
pattern = re.compile("+d{2}\s\d{10}")
result = pattern.sub(
    "",
    "+91 9876543210"
)
print(result.group(0))
print(result.group(1))
print(result.group(2))
```

```python
pattern = re.compile(
    "([^0-9]+)([0-9]{3})-([0-9]{4})-([0-9]{4})"
)
result = pattern.sub(
    r"\1\2-****-\4",
    "John 010-1234-5678"
)
print(result)
```

```python
pattern = re.compile(
    "([^0-9]+)(([0-9]+)[^0-9]+([0-9]+)[^0-9]+([0-9]+))"
)
result = pattern.search("John 010  1234   5678")
print(result.group(3)+"-"+result.group(4)+"-"+result.group(5) )
```

```python
def grep(pattern, text):
    result = []
    compiled = re.compile(pattern)
    for i in range(len(text)):
        if compiled.search(text[i]):
            result.append(text[i])
    return result

pattern = input("Enter pattern: ")
text_list = input("Enter text: ")
text = text_list.split(".")

result = grep(pattern,text)
print(result)
```
